```python
import joblib
import pandas as pd
import numpy as np

model = joblib.load('house_price_model.pkl')  

new_house = pd.DataFrame([{
    'quarter': '2023-Q4',
    'house_type': 'Detached',
    'sales_type': 'Private',
    'area': 'Zealand',
    'region': 'Zealand',
    'year_build': 2005,
    '%_change_between_offer_and_purchase': 3.5,
    'no_rooms': 4,
    'sqm': 120.0,
    'zip_code': 2100
}])


log_price_pred = model.predict(new_house)
price_pred = np.power(10, log_price_pred)  

print(f"Predicted house price: ${price_pred[0]:,.2f}")
```
